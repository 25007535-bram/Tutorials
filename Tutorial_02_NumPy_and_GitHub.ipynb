Q4

```python

random_array = np.random.rand(6, 3)

print("Original array:")
print(random_array)
print("Original shape:", random_array.shape)

transposed_array = random_array.T

print("Transposed array:")
print(transposed_array)
print("Transposed shape:", transposed_array.shape)
```

Q5

```python

n = 2
p = 3
m = 4

A = np.random.rand(n, p)
B = np.random.rand(p, m)

result = np.dot(A, B)

print("Matrix A:")
print(A)
print("Shape of A:", A.shape)

print("Matrix B:")
print(B)
print("Shape of B:", B.shape)

print("Dot product A · B:")
print(result)
print("Shape of result:", result.shape)
```
